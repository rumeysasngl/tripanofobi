```python
import os
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

# 1. Kayıt Klasörlerinin Çalışma Ortamında Oluşturulması
Path("outputs").mkdir(exist_ok=True)

# 2. İkincil Verilerin (5 Hakemli Makalenin İstatistiksel Bulguları) Girişi
yayin_verileri = [
    {"yayin_kodu": "M1_Almahayni_2023", "kadin_fobi_orani": 0.72, "cocukluk_travma_etkisi": 0.85, "genel_prevalans": 0.32},
    {"yayin_kodu": "M2_Al_Raddadi_2024", "kadin_fobi_orani": 0.68, "cocukluk_travma_etkisi": 0.81, "genel_prevalans": 0.28},
    {"yayin_kodu": "M3_McLenon_2019", "kadin_fobi_orani": 0.75, "cocukluk_travma_etkisi": 0.78, "genel_prevalans": 0.35},
    {"yayin_kodu": "M4_McMurtry_2022", "kadin_fobi_orani": 0.65, "cocukluk_travma_etkisi": 0.88, "genel_prevalans": 0.25},
    {"yayin_kodu": "M5_Son_2022", "kadin_fobi_orani": 0.70, "cocukluk_travma_etkisi": 0.83, "genel_prevalans": 0.30}
]

df = pd.DataFrame(yayin_verileri)

# 3. Temel Betimsel İstatistiklerin Hesaplanması ve Özet Tablo Oluşturulması
ozet_df = pd.DataFrame([
    {"metrik": "Ortalama_Kadin_Fobi_Orani", "deger": df["kadin_fobi_orani"].mean()},
    {"metrik": "Ortalama_Cocukluk_Travma_Etkisi", "deger": df["cocukluk_travma_etkisi"].mean()},
    {"metrik": "Maksimum_Genel_Prevalans", "deger": df["genel_prevalans"].max()}
])
# Özet tabloyu kaydetme
ozet_df.to_csv("outputs/tripanofobi_ozet.csv", index=False, encoding="utf-8-sig")

# 4. Matplotlib Kütüphanesi ile Sütun ve Çizgi Grafiği Çizimi
fig, ax = plt.subplots(figsize=(10, 5))
# Sütun Grafiği: Çocukluk travması etkisini karşılaştırır
ax.bar(df["yayin_kodu"], df["cocukluk_travma_etkisi"] * 100, color="crimson", alpha=0.85, label="Çocukluk Travması Etkisi (%)")
# Çizgi Grafiği: Kadın fobi oranındaki yaygınlık eğilimini gösterir
ax.plot(df["yayin_kodu"], df["kadin_fobi_orani"] * 100, color="darkblue", marker="o", linewidth=2, label="Kadın Cinsiyet Bağı (%)")

ax.set_title("Bilimsel Yayınlardaki Tripanofobi Değişken İlişkileri Faktör Dağılımı (İkincil Veri Analizi)")
ax.set_ylabel("Literatür Etki Oranı (%)")
ax.set_xlabel("Referans Alınan Hakemli Yayın Kodları")
ax.legend()
ax.grid(axis="y", alpha=0.2)

# Grafiği outputs klasörüne kalıcı olarak kaydetme
plt.tight_layout()
fig.savefig("outputs/tripanofobi_grafik.png", dpi=160)
plt.show()

print("\n[BAŞARILI] 'outputs/tripanofobi_ozet.csv' ve 'outputs/tripanofobi_grafik.png' dosyaları Colab ortamında üretildi.")
```
